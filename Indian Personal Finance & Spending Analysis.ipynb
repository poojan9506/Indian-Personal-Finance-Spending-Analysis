## 16. Feature Engineering — Total Expenses

A new `Total_Expenses` feature is created by combining the major expense categories available in the dataset.

This makes it easier to analyze overall spending rather than examining each expense category separately.

```python
df['Total_Expenses'] = (
    df['Rent'] +
    df['Loan_Repayment'] +
    df['Insurance'] +
    df['Groceries'] +
    df['Transport'] +
    df['Eating_Out'] +
    df['Entertainment'] +
    df['Healthcare'] +
    df['Education']
)

print("Total_Expenses feature created successfully.")
```

## 33. Correlation Analysis

Correlation measures the strength and direction of a linear relationship between numerical variables.

Values close to **+1** indicate a strong positive relationship, values close to **-1** indicate a strong negative relationship, and values close to **0** indicate a weak linear relationship.

```python
correlation = df.select_dtypes(include=np.number).corr()

print("Correlation Matrix:")
display(correlation)
```

## 34. Correlation Heatmap

The heatmap provides a visual overview of relationships among numerical financial variables.

The analysis focuses on meaningful financial relationships rather than treating correlation as proof of causation.

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    cmap='coolwarm',
    annot=False
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
```

## 35. Key Financial Correlations

The following table focuses on three important variables: Income, Total Expenses, and Desired Savings.

```python
key_correlations = correlation[['Income', 'Total_Expenses', 'Desired_Savings']].sort_values(
    by='Income',
    ascending=False
)

print("Key Financial Correlations:")
display(key_correlations)
```

## 45. Potential Savings Analysis

Potential savings represent the amount that could potentially be saved within each expense category according to the dataset.

## 50. Overall Expense-to-Income Ratio

This metric compares total expenses with total income across all records.

It is calculated as:

**Total Expenses ÷ Total Income × 100**

```python
total_income = df['Income'].sum()
total_expenses = df['Total_Expenses'].sum()

expense_percentage = (total_expenses / total_income) * 100

print(f"Total Income: ₹{total_income:,.2f}")
print(f"Total Expenses: ₹{total_expenses:,.2f}")
print(f"Expenses as a Percentage of Total Income: {expense_percentage:.2f}%")
```

## Key Findings

1. **Rent is the largest expense category** in both total and average spending analysis.
2. **Groceries are the second-largest major expense category** and show the highest potential savings opportunity.
3. Average total expenses generally increase as the number of dependents increases in this dataset.
4. **Tier 1 records have the highest average total expenses**, while Tier 3 records have the lowest.
5. **Tier 3 records have the highest average disposable income** in this dataset.
6. Income and total expenses show a **very strong positive correlation (0.979)**.
7. Income and desired savings show a **strong positive correlation (0.945)**.
8. Only **112 records (0.56%)** have negative disposable income.
9. Groceries show the **highest average potential savings per record**, followed by Transport and Utilities.
10. Across the dataset, total expenses represent approximately **66.38% of total income**.

### Interpretation Note
These findings describe patterns in the provided dataset. They should not automatically be generalized to the entire Indian population, and correlation should not be interpreted as causation.
